```python

def check_prime(num):
    if num <= 1:
        print("Not a prime number")
        return

    if num == 2 or num == 3:
        print("Prime number")
        return

    if num % 2 == 0 or num % 3 == 0:
        print("Not a prime number")
        return

    count = 5
    while count * count <= num:
        if num % count == 0 or num % (count + 2) == 0:
            print("Not a prime number")
            return
        count += 6

    print("Prime number")


def calculate_factorial(num):
    fact = 1

    if num > 0:
        for count in range(1, num + 1):
            fact *= count
        print("Factorial of", num, "is", fact)

    elif num == 0:
        print("Factorial of 0 is 1")

    else:
        print("Please enter a non-negative integer!")


# Function calls
check_prime(17)
calculate_factorial(5)
```

```python

def add(a, b):
    result = a + b
    return result

def sub(a, b):
    result = a - b
    return result

def mul(a, b):
    result = a * b
    return result

def div(a, b):
    if b == 0:
        return "Division by zero is not possible"
    result = a / b
    return result


# Function calls
print("Addition:", add(10, 5))
print("Subtraction:", sub(10, 5))
print("Multiplication:", mul(10, 5))
print("Division:", div(10, 5))
```
